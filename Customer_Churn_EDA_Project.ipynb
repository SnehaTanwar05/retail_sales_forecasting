```python
for column in numerical_columns:

    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    print(f"{column}")

    print(f"Outliers : {len(outliers)}")

    print("-"*40)
```

# Exploratory Data Analysis (EDA)

## Univariate Analysis

Univariate Analysis focuses on analyzing one variable at a time. It helps understand the distribution, frequency, and characteristics of individual variables.

```python
# ==========================================================
# Visualization Settings
# ==========================================================

import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['figure.figsize'] = (10,6)
plt.rcParams['figure.dpi'] = 120

sns.set_style("whitegrid")
sns.set_palette("Set2")
```

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(
    data=df,
    x='churn'
)

plt.title("Customer Churn Distribution", fontsize=16, fontweight='bold')
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

for p in ax.patches:
    ax.annotate(f'{p.get_height():,}',
                (p.get_x()+p.get_width()/2,p.get_height()),
                ha='center',
                va='bottom',
                fontsize=11)

plt.show()
```

### Observation

This chart shows the distribution of customers who stayed and those who churned. It helps determine whether the dataset is balanced or imbalanced.

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df['age'],
    bins=30,
    kde=True,
    color='steelblue'
)

plt.title("Age Distribution of Customers", fontsize=16, fontweight='bold')
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

The histogram illustrates how customer ages are distributed across the dataset. This helps identify the dominant age groups.

```python
plt.figure(figsize=(7,5))

ax = sns.countplot(
    data=df,
    x='gender'
)

plt.title("Gender Distribution", fontsize=16, fontweight='bold')

for p in ax.patches:
    ax.annotate(f'{p.get_height():,}',
                (p.get_x()+p.get_width()/2,p.get_height()),
                ha='center')

plt.show()
```

The dataset contains customers from different gender categories. This visualization helps evaluate whether the dataset is balanced across genders.

```python
plt.figure(figsize=(9,5))

ax = sns.countplot(
    data=df,
    x='subscription_type',
    order=df['subscription_type'].value_counts().index
)

plt.title("Subscription Type Distribution", fontsize=16, fontweight='bold')
plt.xlabel("Subscription Type")
plt.ylabel("Customers")

for p in ax.patches:
    ax.annotate(f'{p.get_height():,}',
                (p.get_x()+p.get_width()/2,p.get_height()),
                ha='center')

plt.show()
```

This chart shows the popularity of different subscription plans among customers.

# Bivariate Analysis

Bivariate Analysis is used to study the relationship between two variables. In this project, we analyze how different customer attributes influence churn.

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='churn',
    y='age'
)

plt.title("Age vs Customer Churn", fontsize=16, fontweight='bold')
plt.xlabel("Churn")
plt.ylabel("Age")

plt.show()
```

This boxplot compares the age distribution of customers who churned and those who remained. Differences in the median and spread may indicate whether age influences churn.

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(
    data=df,
    x='gender',
    hue='churn'
)

plt.title("Gender vs Customer Churn", fontsize=16, fontweight='bold')

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

This chart helps determine whether churn rates differ between gender groups.

```python
plt.figure(figsize=(10,6))

ax = sns.countplot(
    data=df,
    x='subscription_type',
    hue='churn'
)

plt.title("Subscription Type vs Churn", fontsize=16, fontweight='bold')

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

This visualization compares churn across subscription plans. Plans with higher churn rates may require business attention.

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='churn',
    y='tenure_months'
)

plt.title("Customer Tenure vs Churn", fontsize=16, fontweight='bold')

plt.show()
```

Customers with shorter tenure often exhibit higher churn. Compare the medians and quartiles between the two groups.

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='churn',
    y='avg_order_value'
)

plt.title("Average Order Value vs Churn", fontsize=16, fontweight='bold')

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='churn',
    y='total_orders'
)

plt.title("Total Orders vs Churn", fontsize=16, fontweight='bold')

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='churn',
    y='support_tickets'
)

plt.title("Support Tickets vs Churn", fontsize=16, fontweight='bold')

plt.show()
```

If churned customers tend to have more support tickets, it may indicate dissatisfaction with products or services.

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='churn',
    y='last_purchase_days_ago'
)

plt.title("Last Purchase Days vs Churn", fontsize=16, fontweight='bold')

plt.show()
```

```python
subscription_churn = pd.crosstab(
    df['subscription_type'],
    df['churn'],
    normalize='index'
) * 100

subscription_churn
```

```python
subscription_churn.plot(
    kind='bar',
    stacked=True,
    figsize=(10,6)
)

plt.title("Churn Percentage by Subscription Type", fontsize=16, fontweight='bold')
plt.ylabel("Percentage")

plt.legend(title='Churn')
plt.show()
```

```python
plt.figure(figsize=(10,8))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title("Correlation Matrix", fontsize=16, fontweight='bold')

plt.show()
```

## 6.1 Correlation Matrix

A correlation matrix measures the strength and direction of the relationship between numerical variables.

```python
# ==========================================================
# Correlation Matrix
# ==========================================================

correlation_matrix = df.corr(numeric_only=True)

correlation_matrix
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Correlation Heatmap", fontsize=16, fontweight='bold')

plt.show()
```

### Observation

```python
sns.pairplot(
    df[['age',
        'tenure_months',
        'avg_order_value',
        'total_orders',
        'support_tickets',
        'churn']],
    hue='churn',
    corner=True
)

plt.show()
```

### Observation
